# 쌍체 표본 t검정

- 다이어트 전과 후의 차이가 유의한지 검정

## 가설 설정

- 귀무가설 : 다이어트 전, 후 차이가 없다.   (==)
- 대립가설 : 다이어트 전, 후 차이가 있다.   (!=)

```python
import pandas as pd
```

```python
df = pd.read_csv("datas2/다이어트약_효과검증.csv", encoding='utf-8')
df.columns
```

```python
# A그룹
before = df['다이어트전(kg)'] # 열 내용 바로 access 가능
# B그룹
after = df['다이어트후(kg)'] # 열 내용 바로 access 가능

print(before)
print(after)
```

# 정규성 검정

- 표본이 20개 이상이므로 d'agostino 방식 사용.

    - 영가설 : 다이어트 전과 후의 차이 d는 정규 분포를 따른다.
    - 대립가설 : 다이어트 전과 후의 차이 d는 정규 분포를 따르지 않는다. 

```python
from scipy import stats
```

```python
d = before - after
print(stats.normaltest(d))
#statistic, p_value = stats.normaltest(d)
#print(statistic)
#print(p_value)
```

- p_value >= 0.05 이므로 영가설 수용.
- 다이어트 전과 후의 차이 d는 정규 분포를 따른다고 볼 수 있다는 것이 확률적으로 유의미함. 

- 참고용: shapiro-wilk 검정 방식

```python
stats.shapiro(after - before)
```

- 참고용 : 실제 평균은 차이가 있는가

# 쌍체 표본 t 검정

- 영가설 : 다이어트 전 후 사이의 결과가 같아 차이가 없다. 효과가 없다. (==)
- 대립가설 : 다이어트 전 후 사이의 결과가 같지 않아 차이가 있다. 효과가 있다. (!=)

```python
print(stats.ttest_rel(before, after)) # before, after는 길이가 같아야 함.
```

- 주의! e-13이므로 p_value 크기가 매우 작음을 알 수 있다.
- p_value < 0.05 이므로 영가설 기각. 다이어트 전과 후가 차이가 존재함이 통계적으로 유의미함.
- 즉, 다이어트 전후 차이가 존재하므로 다이어트는 유의미한 결과를 남김.
- 통계량이 양수 이기 때문에 양의 효과 있다는 것을 알 수 있음

- 참고용 : 대응표본 t통계량 공식에 맞춰서 계산

```python
# 전후 차이 계산(개인별 체중 차이)
diff = before - after

# 체중 차이의 평균 계산(평균 체중 감소량)
mean_diff = np.mean(diff)

# 체중 차이의 표준편차(감소량의 표준편차)
std_diff = np.std(diff, ddof=1)

# 체중 차이의 표준오차(표본에서 구한 평균 체중 감소량의 불확실성을 나타냄)
standard_error = std_diff / np.sqrt(len(diff))

# t-통계량 계산(평균 체중 감소량이 표준오차보다 몇 배 큰지를 나타냄)
t_statistic = mean_diff / standard_error
t_statistic
```

```python
# 계산 결과 출력
print(mean_diff)       # 평균 체중 감소량
print(std_diff)        # 체중 감소량의 표준편차
print(standard_error)  # 평균 체중 감소량의 표준오차
print(t_statistic)     # t 통계량
```
